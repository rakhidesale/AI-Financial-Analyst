#### CRISP-DM Phase

Data Understanding

```python
# Importing Required Libraries

import os
import json
from pathlib import Path

import pandas as pd

print("Libraries imported successfully.")
```

#### Step 1 - Locating the EDGAR Dataset

The historical EDGAR dataset is stored locally in the project directory.

The dataset is organised by reporting year, with each folder containing
multiple JSON files representing SEC filings.

```python
# Locate Dataset

DATASET_PATH = Path("../data/edgar")

print("Dataset Location:")
print(DATASET_PATH.resolve())
```

#### Step 2 - Explore Folder Structure

The folder structure is inspected to confirm that all reporting years are
available before analysing individual filings.

```python
# Explore Dataset Structure

years = sorted(
    [
        folder.name
        for folder in DATASET_PATH.iterdir()
        if folder.is_dir()
    ]
)

print("=" * 60)
print("AVAILABLE YEARS")
print("=" * 60)

for year in years:
    print(year)
```

#### Step 3 - Counting Files

The number of JSON files in each reporting year is calculated to gain an
understanding of the dataset size.

```python
# Count JSON File

summary = []

for year in years:

    files = list((DATASET_PATH / year).glob("*.json"))

    summary.append(
        {
            "Year": year,
            "Number of Files": len(files)
        }
    )

summary_df = pd.DataFrame(summary)

display(summary_df)
```

#### Step 4 - Selecting a Sample Filing

A single JSON filing is selected to inspect the dataset structure before
processing the remaining files.

```python
# Select Sample File

sample_file = sorted(
    (DATASET_PATH / "2018").glob("*.json")
)[0]

print(sample_file)
```

#### Step 5 - Loading Sample Filing
The selected JSON file is loaded into Python for inspection.

```python
# Loading JSON File

with open(sample_file, "r", encoding="utf-8") as f:

    filing = json.load(f)

print("JSON loaded successfully.")
```

#### Step 6 - Inspecting JSON Structure

The top-level keys of the JSON object are displayed to understand the
dataset schema.

```python
# Displaying JSON Keys

print(type(filing))

print()

print(filing.keys())
```

#### Step 7 - Display Filing Metadata

The metadata associated with the selected filing is displayed to verify
the reporting year and company identifier.

```python
# Display Filing Metadata

print("=" * 60)
print("FILING METADATA")
print("=" * 60)

print(f"Filename : {filing['filename']}")
print(f"CIK      : {filing['cik']}")
print(f"Year     : {filing['year']}")
```

#### Step 8 - Verify Required Narrative Sections

The project requires three narrative sections from each filing:

- Section 1 (Business)
- Section 1A (Risk Factors)
- Section 7 (Management Discussion & Analysis)

The presence of these sections is verified before proceeding.

```python
# Verify Required Sections

required_sections = [
    "section_1",
    "section_1A",
    "section_7"
]

print("=" * 60)
print("SECTION AVAILABILITY")
print("=" * 60)

for section in required_sections:

    exists = section in filing

    print(f"{section:<12}: {exists}")
```

```python
# Section Length

print("=" * 60)
print("SECTION LENGTH")
print("=" * 60)

for section in required_sections:

    text = filing.get(section, "")

    print(f"{section:<12}: {len(text):,} characters")
```

```python
# Preview Sections

for section in required_sections:

    print("=" * 60)
    print(section.upper())
    print("=" * 60)

    preview = filing[section][:500]

    print(preview)
    print("\n")
```

#### Conclusion

The historical EDGAR dataset was successfully validated.

The validation confirmed that:

- The dataset can be loaded successfully.
- Each filing contains the company CIK and reporting year.
- The required narrative sections (Section 1, Section 1A and Section 7) are available.
- The dataset is suitable for the Retrieval-Augmented Generation (RAG) pipeline.

Based on this validation, the historical EDGAR dataset will be used as the narrative data source for the remainder of the project.
